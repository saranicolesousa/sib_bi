# SIB - Exercício 6.2

Teste da função `stratified_train_test_split` com o dataset iris.

```python
import numpy as np

from si.io.csv_file import read_csv
from si.model_selection.split import train_test_split, stratified_train_test_split
```

```python
iris = read_csv('../datasets/iris/iris.csv', sep=',', features=True, label=True)

print('dimensões:', iris.shape())
print('distribuição das classes:', dict(zip(*np.unique(iris.y, return_counts=True))))
```

## Divisão estratificada

```python
train, test = stratified_train_test_split(iris, test_size=0.2, random_state=42)

print('treino:', train.shape(), dict(zip(*np.unique(train.y, return_counts=True))))
print('teste :', test.shape(), dict(zip(*np.unique(test.y, return_counts=True))))
```

Cada classe fica com 40 amostras no treino e 10 no teste, isto é, a proporção original de 1/3 por classe é mantida nos dois conjuntos.

## Comparação com a divisão aleatória simples

```python
print('classe [setosa, versicolor, virginica] no conjunto de teste\n')
for seed in range(6):
    _, test_random = train_test_split(iris, test_size=0.2, random_state=seed)
    _, test_strat = stratified_train_test_split(iris, test_size=0.2, random_state=seed)
    counts_random = [int(v) for v in np.unique(test_random.y, return_counts=True)[1]]
    counts_strat = [int(v) for v in np.unique(test_strat.y, return_counts=True)[1]]
    print(f'seed {seed}: aleatória {counts_random}   estratificada {counts_strat}')
```

A divisão aleatória produz conjuntos de teste desequilibrados: com a seed 4, por exemplo, ficam 16 setosas e apenas 5 versicolor. A estratificada devolve sempre 10/10/10, independentemente da semente.

Isto importa porque a métrica de avaliação é dominada pelas classes mais representadas no conjunto de teste. Num dataset desequilibrado, uma classe rara pode até ficar totalmente de fora do teste com uma divisão aleatória.
